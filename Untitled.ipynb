```python
import re
import random
import pickle
import numpy as np
from collections import defaultdict, Counter
from scipy.stats import dirichlet
import os
import pandas as pd

def load_pubtator(path):
    """Loads documents from a PubTator formatted file."""
    docs = []
    cur = {"pmid": None, "text": "", "annotations": []}
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip()
            if line == "":
                if cur["pmid"]:
                    docs.append(cur)
                cur = {"pmid": None, "text": "", "annotations": []}
                continue
            if "|t|" in line or "|a|" in line:
                pmid, tag, txt = line.split("|", 2)
                if tag == "t":
                    cur["pmid"], cur["text"] = pmid, txt
                else:
                    cur["text"] += " " + txt
            else:
                parts = line.split("\t")
                if len(parts) >= 6:
                    _, start, end, mention, semtypes, _ = parts[:6]
                    cur["annotations"].append({
                        "start": int(start),
                        "end": int(end),
                        "mention": mention,
                        "semtype": semtypes.split(",")[0]
                    })
    return docs

def read_pmids(path):
    """Reads a list of PMIDs from a file."""
    with open(path, encoding="utf-8") as f:
        return [line.strip() for line in f if line.strip()]

def docs_to_sentences(docs):
    """Converts documents with annotations into tokenized sentences with IOB2 tags."""
    pattern = re.compile(r"[^.?!]+[.?!]?")
    sents = []
    for d in docs:
        for m in pattern.finditer(d["text"]):
            sent = m.group().strip()
            if not sent:
                continue
            toks = sent.split()
            tags = ["O"] * len(toks)
            # Sort annotations by start position to handle nested entities correctly if needed
            sorted_annotations = sorted(d["annotations"], key=lambda x: x['start'])
            for ann in sorted_annotations:
                ment = ann["mention"].split()
                for i in range(len(toks) - len(ment) + 1):
                    # Check for a match and ensure the position is not already tagged
                    if toks[i:i+len(ment)] == ment and all(tags[j] == "O" for j in range(i, i + len(ment))):
                        tags[i] = "B-" + ann["semtype"]
                        for j in range(1, len(ment)):
                            tags[i+j] = "I-" + ann["semtype"]
            sents.append({"tokens": toks, "labels": tags})
    return sents

def multi_label_dirichlet_routing_partition(sentences, num_clients, alpha=0.1, seed=42):
    """
    Partitions sentences among clients using a multi-label Dirichlet routing strategy.
    This creates a Non-IID distribution based on the labels present in each sentence.
    """
    random.seed(seed)
    np.random.seed(seed)

    # 1. For each label type, generate a Dirichlet distribution probability vector
    #    This vector determines how likely a label is to be assigned to each client.
    all_labels = sorted({lab[2:] for s in sentences for lab in s["labels"] if lab != "O"})
    label2client_probs = {
        l: np.random.dirichlet([alpha] * num_clients) for l in all_labels
    }

    # 2. Iterate through each sentence and decide its client assignment based on its labels.
    client_sent_ids = {i: [] for i in range(num_clients)}
    for idx, s in enumerate(sentences):
        labels_in_sent = list(set([lab[2:] for lab in s["labels"] if lab != "O"]))
        
        # If the sentence has no labels, assign it to a random client.
        if not labels_in_sent:
            assigned_client = random.randint(0, num_clients - 1)
        else:
            # For sentences with labels, calculate a score for each client.
            # The score is the sum of probabilities for each label present in the sentence.
            scores = np.zeros(num_clients)
            for l in labels_in_sent:
                if l in label2client_probs:
                    scores += label2client_probs[l]
            # Assign the sentence to the client with the highest score.
            assigned_client = int(np.argmax(scores))
            
        client_sent_ids[assigned_client].append(idx)

    return {cid: [sentences[i] for i in sent_ids] for cid, sent_ids in client_sent_ids.items()}

def load_and_split_pubtator(
    pubtator_path,
    trng_pmids_path,
    dev_pmids_path,
    test_pmids_path,
    num_clients=10,
    partition_strategy="noniid",
    alpha=0.01,
    seed=42,
    cache_path=None
):
    """Main function to load, process, and partition the PubTator dataset."""
    if cache_path and os.path.exists(cache_path):
        print(f"Loading cached data from {cache_path}...")
        with open(cache_path, "rb") as f:
            return pickle.load(f)

    print("Loading and processing data from source...")
    docs = load_pubtator(pubtator_path)
    pmid_map = {d["pmid"]: d for d in docs}

    # Read PMIDs for each split
    train_docs = [pmid_map[p] for p in read_pmids(trng_pmids_path) if p in pmid_map]
    dev_docs   = [pmid_map[p] for p in read_pmids(dev_pmids_path)   if p in pmid_map]
    test_docs  = [pmid_map[p] for p in read_pmids(test_pmids_path)  if p in pmid_map]

    # Convert documents to sentences
    train_sents = docs_to_sentences(train_docs)
    dev_sents   = docs_to_sentences(dev_docs)
    test_sents  = docs_to_sentences(test_docs)
    
    print(f"Total training sentences: {len(train_sents)}")

    # Partition the training data according to the specified strategy
    if partition_strategy == "noniid":
        print(f"Partitioning training data into {num_clients} clients with non-iid strategy (alpha={alpha})...")
        client_train = multi_label_dirichlet_routing_partition(train_sents, num_clients, alpha, seed)
    else:
        raise ValueError(f"This script is designed for 'noniid' strategy. Got: {partition_strategy}")

    # Convert client data from dict to a list of lists for easier access
    client_train_sents = [client_train.get(i, []) for i in range(num_clients)]
    
    # For simplicity, we are only focusing on the training set partition here.
    # The original function also partitions dev/test sets.
    result = (client_train_sents, dev_sents, test_sents)

    if cache_path:
        os.makedirs(os.path.dirname(cache_path), exist_ok=True)
        with open(cache_path, "wb") as f:
            pickle.dump(result, f)

    return result

def print_client_data_distribution(client_sents):
    """Prints the number of sentences and label distribution for each client."""
    print("\n" + "="*50)
    print("      Client Data Distribution Analysis")
    print("="*50)
    
    total_sentences = 0
    for i, sents in enumerate(client_sents):
        num_sents = len(sents)
        total_sentences += num_sents
        print(f"\n--- Client {i}: {num_sents} sentences ---")
        
        if not sents:
            print("This client has no data.")
            continue
            
        labs = [lab[2:] for s in sents for lab in s["labels"] if lab != "O"]
        dist = Counter(labs)
        
        # Create a DataFrame for nice printing
        if dist:
            df = pd.DataFrame(dist.items(), columns=["Label", "Count"]).sort_values("Count", ascending=False)
            print("Top 10 most frequent labels:")
            print(df.head(10).to_string(index=False))
            print(f"Total unique labels for this client: {len(dist)}")
        else:
            print("No labels found for this client.")
            
    print("\n" + "="*50)
    print(f"Total sentences across all clients: {total_sentences}")
    print("="*50)


# ==============================================================================
# Main Execution Block
# ==============================================================================
if __name__ == "__main__":
    # --- Configuration ---
    # PLEASE REPLACE THESE PATHS WITH YOUR ACTUAL FILE PATHS
    # NOTE: You need to create dummy files for this script to run if you don't have the data.
    # For example, create empty .txt files with these names.
    PUBTATOR_PATH = "data_set/corpus_pubtator.txt"
    TRNG_PMIDS_PATH = "data_set/corpus_pubtator_pmids_trng.txt"
    DEV_PMIDS_PATH = "data_set/corpus_pubtator_pmids_dev.txt"
    TEST_PMIDS_PATH = "data_set/corpus_pubtator_pmids_test.txt"

    # Check if dummy files need to be created
    for path in [PUBTATOR_PATH, TRNG_PMIDS_PATH, DEV_PMIDS_PATH, TEST_PMIDS_PATH]:
        if not os.path.exists(path):
            print(f"Creating dummy file: {path}")
            with open(path, "w") as f:
                # Add a minimal amount of data to prevent errors
                if "pubtator.txt" in path:
                    f.write("12345|t|Sample title.\n")
                    f.write("12345|a|Sample abstract.\n")
                    f.write("12345\t0\t6\tSample\tT058\tC00000\n\n")
                else:
                    f.write("12345\n")


    # --- Parameters for Partitioning ---
    NUM_CLIENTS = 10
    NONIID_ALPHA = 0.01
    SEED = 42

    # --- Load and Partition Data ---
    client_train_data, _, _ = load_and_split_pubtator(
        pubtator_path=PUBTATOR_PATH,
        trng_pmids_path=TRNG_PMIDS_PATH,
        dev_pmids_path=DEV_PMIDS_PATH,
        test_pmids_path=TEST_PMIDS_PATH,
        num_clients=NUM_CLIENTS,
        partition_strategy="noniid",
        alpha=NONIID_ALPHA,
        seed=SEED
    )

    # --- Print the Distribution ---
    print_client_data_distribution(client_train_data)
```
